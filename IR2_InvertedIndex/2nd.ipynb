```python

import string

# Step 1: Create a text file
with open("file.txt", "w") as f:
    f.write("This is the first word.\n")
    f.write("This is the second text Hello How are you?\n")
    f.write("This is the third this is it now.")

# Step 2: Read the file
with open("file.txt", "r") as f:
    lines = f.readlines()

# Step 3: Define stop words
stop_words = ["is", "the", "are", "how", "you", "it"]

# Step 4: Create inverted index
inverted_index = {}

for line_no, line in enumerate(lines, start=1):

    # Convert to lowercase
    line = line.lower()

    # Remove punctuation
    line = line.translate(str.maketrans("", "", string.punctuation))

    # Tokenization
    words = line.split()

    # Remove stop words
    for word in words:
        if word not in stop_words:

            if word not in inverted_index:
                inverted_index[word] = []

            inverted_index[word].append(line_no)

# Step 5: Display inverted index
print("Inverted Index:")

for word, line_numbers in inverted_index.items():
    print(word, ":", line_numbers)

# Step 6: Search for a word
search_word = input("\nEnter word to search: ").lower()

if search_word in inverted_index:
    print("Word found in lines:", inverted_index[search_word])
else:
    print("Word not found")
```
