```python
import pandas as pd
import numpy as np

from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.estimators import MaximumLikelihoodEstimator
from pgmpy.inference import VariableElimination

# Step 1: Read Heart Disease Dataset
heartDisease = pd.read_csv("heart.csv")

# Step 2: Handle missing values
heartDisease = heartDisease.replace("?", np.nan)
heartDisease = heartDisease.dropna()

# Step 3: Select required attributes
heartDisease = heartDisease[
    ['age', 'sex', 'exang', 'cp',
     'target', 'restecg', 'chol']
]

# Step 4: Create Bayesian Network
model = DiscreteBayesianNetwork([
    ('age', 'target'),
    ('sex', 'target'),
    ('exang', 'target'),
    ('cp', 'target'),
    ('target', 'restecg'),
    ('target', 'chol')
])

# Step 5: Learn CPDs using Maximum Likelihood Estimator
print("Learning CPD using Maximum Likelihood Estimator")

model.fit(heartDisease)

# Step 6: Perform inference
print("\nInferencing with Bayesian Network:")

HeartDisease_infer = VariableElimination(model)

# Step 7: Probability of Heart Disease given restecg
print("\n1. Probability of Heart Disease given restecg = 1")

q1 = HeartDisease_infer.query(
    variables=['target'],
    evidence={'restecg': 1}
)

print(q1)

# Step 8: Probability of Heart Disease given chest pain
print("\n2. Probability of Heart Disease given cp = 2")

q2 = HeartDisease_infer.query(
    variables=['target'],
    evidence={'cp': 2}
)

print(q2)
```
