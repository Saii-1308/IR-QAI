```python

import numpy as np
from bs4 import BeautifulSoup

# Step 1: Create sample web pages
pages = {
    "A": '<a href="B">Page B</a><a href="C">Page C</a>',
    "B": '<a href="C">Page C</a>',
    "C": '<a href="A">Page A</a>'
}

# Step 2: Extract links using BeautifulSoup
links = {}

for page, html in pages.items():
    soup = BeautifulSoup(html, "html.parser")
    links[page] = [
        tag.get("href")
        for tag in soup.find_all("a")
    ]

print("Web Page Links:")
for page in links:
    print(page, ":", links[page])

# Step 3: Create transition matrix
n = len(pages)
page_names = list(pages.keys())

matrix = np.zeros((n, n))

for i, page in enumerate(page_names):
    outgoing = links[page]

    if len(outgoing) == 0:
        matrix[i] = 1 / n
    else:
        for link in outgoing:
            j = page_names.index(link)
            matrix[i][j] = 1 / len(outgoing)

print("\nTransition Matrix:")
print(matrix)

# Step 4: Initialize PageRank
rank = np.ones(n) / n

# Step 5: Set damping factor
d = 0.85

# Step 6: Apply PageRank using Power Method
for iteration in range(100):
    new_rank = (
        d * np.dot(matrix.T, rank)
        + (1 - d) / n
    )

    if np.linalg.norm(new_rank - rank, 1) < 0.0001:
        break

    rank = new_rank

# Step 7: Display PageRank
print("\nPageRank Values:")

for i, page in enumerate(page_names):
    print(page, ":", round(rank[i], 4))

# Step 8: Display important page
important_page = page_names[np.argmax(rank)]

print("\nMost Important Page:", important_page)
```
